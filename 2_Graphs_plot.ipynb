```python
import numpy as np
import pandas as pd
from nilearn import plotting
from nilearn import datasets
import matplotlib.pyplot as plt

# --- 1. CONFIGURACIÓN INICIAL Y CARGA DE DATOS ---

# Rutas a los archivos de las matrices en formato .csv
file_paths = {
    'Alzheimer': 'matrices/matriz_AD.csv',
    'Control Alzheimer': 'matrices/matriz_CNAD.csv',
    'DCL': 'matrices/matriz_MCI.csv',
    'Control DCL': 'matrices/matriz_CNMCI.csv'
}

# Cargar las matrices desde los archivos .csv
try:
    # header=None asume que el CSV no tiene una fila de encabezado
    correlation_matrices = {name: pd.read_csv(path, header=0).values for name, path in file_paths.items()}
    print(" Matrices cargadas exitosamente desde archivos CSV.")
except FileNotFoundError as e:
    print(f" Error: No se pudo encontrar el archivo {e.filename}.")
    print("Por favor, verifica que la ruta y la estructura de carpetas ('matrices/') sean correctas.")
    exit()

# --- 2. OBTENCIÓN DE COORDENADAS Y ETIQUETAS DEL ATLAS SEITZMAN ---

print(" Descargando y cargando coordenadas del atlas Seitzman 2018...")
# Obtener el atlas directamente con la función de nilearn (¡ACTUALIZADO!)
seitzman_atlas = datasets.fetch_coords_seitzman_2018()

# Extraer las coordenadas como un array de NumPy
node_coords = np.vstack((
    seitzman_atlas.rois['x'],
    seitzman_atlas.rois['y'],
    seitzman_atlas.rois['z']
)).T

# Usar las etiquetas reales del atlas 
node_labels = seitzman_atlas.networks
print("Coordenadas y etiquetas del atlas listas.")

# --- 3. FUNCIÓN PARA VISUALIZACIÓN ---

def visualizar_conectividad(matrix, coords, labels, title):
    """
    Genera y guarda visualizaciones 3D de una matriz de conectividad.
    
    Args:
        matrix (np.ndarray): Matriz de correlación (ej. 300x300).
        coords (np.ndarray): Coordenadas de los nodos (ROIs).
        labels (list): Nombres de las etiquetas de los nodos.
        title (str): Título para los gráficos y nombre del archivo.
    """
    
    print(f" Generando visualizaciones para: {title}...")
    
    # Umbral: visualizamos solo el 2% de las conexiones más fuertes para mayor claridad
    threshold = f"{100 - 2}%"  #%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%% CAMBIADO AQUÍ %%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
    
    # --- Visualización 3D: Conectoma en el Cerebro ---
    # 
    fig_3d = plt.figure(figsize=(10, 8))
    plotting.plot_connectome(
        adjacency_matrix=matrix,
        node_coords=coords,
        edge_threshold=threshold,
        title=f'Conectividad 3D - {title}',
        display_mode='x', # Vista axial #%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%% CAMBIADO AQUÍ %%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%%
        figure=fig_3d
    )
    fig_3d.savefig(f'conectoma_3D_{title.replace(" ", "_")}.png', dpi=300)
    plt.close(fig_3d) 
      
# --- 4. BUCLE PRINCIPAL PARA GENERAR TODAS LAS VISUALIZACIONES ---

for group_name, matrix in correlation_matrices.items():
    visualizar_conectividad(matrix, node_coords, node_labels, group_name)

print("\n ¡Proceso completado! Revisa la carpeta para ver tus gráficos.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

def plot_circular_connectome(matrix, labels, title):
    """
    Generates a 2D circular visualization of a connectivity matrix.
    
    Args:
        matrix (np.ndarray): The connectivity matrix.
        labels (list): Names of the nodes.
        title (str): Title for the plot.
    """
    print(f" Generating circular plot for: {title}...")
    
    # Create a graph from the adjacency matrix
    G = nx.from_numpy_array(np.abs(matrix))

    # Set up the circular layout
    pos = nx.circular_layout(G)
    
    plt.figure(figsize=(10, 10))
    plt.title(title, fontsize=16)

    # Draw nodes and labels
    nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=100)
    nx.draw_networkx_labels(G, pos, labels=dict(zip(range(len(labels)), labels)), font_size=8)
    
    # Draw edges with color and thickness based on weight
    edges = G.edges(data=True)
    weights = [abs(d['weight']) for u, v, d in edges]
    
    nx.draw_networkx_edges(G, pos, width=weights, edge_color='gray', alpha=0.5)
    
    plt.axis('off') # Hide axes
    plt.savefig(f'circular_conectome_{title.replace(" ", "_")}.png', dpi=300)
    plt.close()

# You would call this function in your main loop like this:
for group_name, matrix in correlation_matrices.items():
    # Note: node_labels should be a list of strings, not a DataFrame
    plot_circular_connectome(matrix, node_labels, group_name)
```
